```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Given reward patterns
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Part B - Initialize N, R and Q
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

# Keep track of which reward to use next
reward_index = {'A': 0, 'B': 0, 'C': 0}

total_reward = 0

print("Epsilon-Greedy Multi-Armed Bandit")
print("----------------------------------")

# Play 10 times
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Part C - Epsilon-Greedy Decision
    if r < epsilon:
        # Exploration
        selected = random.choice(machines)
        decision = "Exploration"

    else:
        # Exploitation
        max_q = max(Q.values())

        # Handle ties
        best_machines = [
            m for m in machines
            if Q[m] == max_q
        ]

        selected = random.choice(best_machines)
        decision = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Random number = {r:.3f}")
    print(f"Decision = {decision}")
    print(f"Selected Machine = {selected}")

    # Part D - Get reward from given reward pattern
    index = reward_index[selected]

    # Use reward pattern cyclically
    reward = reward_patterns[selected][index]

    # Move to next reward
    reward_index[selected] = (
        reward_index[selected] + 1
    ) % len(reward_patterns[selected])

    print(f"Reward obtained = {reward}")

    # Part E - Update N, R and Q

    # Increase number of times machine was played
    N[selected] += 1

    # Add reward to total reward of that machine
    R[selected] += reward

    # Incremental sample-average update
    Q[selected] = Q[selected] + (
        (reward - Q[selected]) / N[selected]
    )

    # Update total reward
    total_reward += reward

    print(f"N({selected}) = {N[selected]}")
    print(f"R({selected}) = {R[selected]}")
    print(f"Q({selected}) = {Q[selected]:.2f}")


# Part F - Display final results

print("\n================================")
print("FINAL RESULTS")
print("================================")

print("\nMachine\tN\tR\tQ")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]}\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward after 10 plays =", total_reward)

# Best machine according to final Q
best_machine = max(Q, key=Q.get)

print("Best Machine based on final Q =", best_machine)
```

```python
import gymnasium as gym


def random_agent(environment_name):

    # 1. Create the selected environment
    env = gym.make(environment_name)

    # 2. Reset the environment
    # 3. Observe the initial state
    state, info = env.reset()

    total_reward = 0
    steps = 0

    print("\n===================================")
    print("Environment:", environment_name)
    print("Initial State:", state)
    print("===================================")

    # Continue until the episode terminates
    while True:

        # 4. Identify available actions
        print("Available Actions:", env.action_space)

        # 5. Select an action randomly
        action = env.action_space.sample()

        # 6. Execute the action in the environment
        next_state, reward, terminated, truncated, info = env.step(action)

        # 7. Observe next state and reward
        print("Action:", action)
        print("Next State:", next_state)
        print("Reward:", reward)

        # Update values
        total_reward += reward
        steps += 1

        # Move to next state
        state = next_state

        # 8. Continue until episode terminates
        if terminated or truncated:
            break

    # 9. Calculate average reward
    average_reward = total_reward / steps

    print("\n========== RESULTS ==========")
    print("Total Reward:", total_reward)
    print("Number of Steps:", steps)
    print("Average Reward:", round(average_reward, 2))

    env.close()


# Run Random Agent

random_agent("CartPole-v1")
random_agent("MountainCar-v0")
random_agent("FrozenLake-v1")
```

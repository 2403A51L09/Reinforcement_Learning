TAXI - ENVIRONMENT

```python
import gymnasium as gym

env = gym.make("Taxi-v4")
print("environment created")

current_state, info = env.reset()
print("Initial State ID:", current_state)

taxi_row, taxi_col, pickup, destination = env.unwrapped.decode(current_state)

print(f"Taxi Row: {taxi_row}, Taxi Col: {taxi_col}")
print(f"Pickup (Passenger Location): {pickup}")
print(f"Destination: {destination}")
```

```python
print("Action space:", env.action_space)

action = env.action_space.sample()

next_state, reward, terminated, truncated, info = env.step(action)

print(f"Current State: {current_state}")
print(f"Action Taken: {action}")
print(f"Next State: {next_state}")
print(f"Reward: {reward}")
print(f"Terminated: {terminated}, Truncated: {truncated}")

t_row, t_col, p_loc, dest = env.unwrapped.decode(next_state)
print(f"Decoded Next State -> Row: {t_row}, Col: {t_col}, Pickup: {p_loc}, Destination: {dest}")
```

```python
# import gymnasium as gym

# env = gym.make("Taxi-v4")

current_state, info = env.reset()

num_episodes = 10
total_reward = 0

for episode in range(num_episodes):

    action = env.action_space.sample()

    next_state, reward, terminated, truncated, info = env.step(action)

    total_reward += reward

    print(f"Episode: {episode + 1} | Action: {action} | Current State: {current_state} | "
          f"Next State: {next_state} | Reward: {reward} | Terminated: {terminated} | Truncated: {truncated}")

    if terminated or truncated:
        break

    current_state = next_state

print("--------------------------------------------------")
print("Total Reward:", total_reward)
print("Average Reward:", total_reward / (episode + 1))

env.close()
```

CLIFFWALKING - ENVIRONMENT

```python
import gymnasium as gym

env = gym.make("CliffWalking-v1")

state, info = env.reset()
print(f"Current State: {state}")

row = state // 12
col = state % 12
print(f"Row = {row}, Column = {col}")
```

```python
obs_space = env.observation_space
print("Observation Space:", obs_space)

action_space = env.action_space
print("Action Space:", action_space)

actions = {
    0: "UP",
    1: "RIGHT",
    2: "DOWN",
    3: "LEFT"
}

for i in range(env.action_space.n):
    print(f"Action {i}: {actions[i]}")
```

```python
for ep in range(10):
    random_action = env.action_space.sample()
    print(f"Episode {ep}: Action {random_action} -> {actions[random_action]}")
```

```python
# import gymnasium as gym

# env = gym.make("CliffWalking-v1")
episodes = 10

for ep in range(episodes):
    state, info = env.reset()
    total_reward = 0
    terminated = False

    while not terminated:
        action = env.action_space.sample()

        next_state, reward, terminated, truncated, info = env.step(action)

        total_reward += reward

        if truncated:
            break

    print(f"Total Reward for Episode {ep}: {total_reward}")

env.close()
```

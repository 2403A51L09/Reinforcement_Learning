```python
import random

# ==========================================
# Part A - Machine Initialization
# ==========================================

machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2


# ==========================================
# Part B - Initialize N, R and Q
# ==========================================

N = {'A': 0, 'B': 0, 'C': 0}       # Participation count
R = {'A': 0, 'B': 0, 'C': 0}       # Total reward
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0} # Estimated value

total_reward = 0


print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("--------------------------------------------")


# ==========================================
# Part C - Modified ε-Greedy Decision
# ==========================================

for play in range(1, N_play + 1):

    # Generate random number between 0 and 1
    r = random.random()

    # Find maximum Q-value
    max_q = max(Q.values())

    # Find all machines having maximum Q
    highest_machines = [
        m for m in machines
        if Q[m] == max_q
    ]

    # --------------------------------------
    # Exploration
    # --------------------------------------

    if r < epsilon:

        # If all machines have the same Q,
        # there is no unique highest machine.
        if len(highest_machines) == len(machines):

            selected = random.choice(machines)

        else:
            # Exclude highest-Q machine
            exploration_machines = [
                m for m in machines
                if m not in highest_machines
            ]

            selected = random.choice(exploration_machines)

        decision = "Exploration"

    # --------------------------------------
    # Exploitation
    # --------------------------------------

    else:

        # If there is a tie, randomly choose
        # one of the highest-Q machines
        selected = random.choice(highest_machines)

        decision = "Exploitation"


    # ==========================================
    # Part D - Reward Collection
    # ==========================================

    print(f"\nPlay {play}")
    print(f"Random number = {r:.3f}")
    print(f"Decision = {decision}")
    print(f"Selected Machine = {selected}")

    reward = float(
        input(f"Enter reward for Machine {selected}: ")
    )


    # ==========================================
    # Part E - Q-Value Update
    # ==========================================

    # Increase participation count
    N[selected] += 1

    # Add reward to total reward of machine
    R[selected] += reward

    # Incremental sample-average formula
    Q[selected] = Q[selected] + (
        (reward - Q[selected]) / N[selected]
    )

    # Update total reward
    total_reward += reward


    print(f"N({selected}) = {N[selected]}")
    print(f"R({selected}) = {R[selected]}")
    print(f"Q({selected}) = {Q[selected]:.2f}")


# ==========================================
# Part F - Display Final Results
# ==========================================

print("\n============================================")
print("FINAL RESULTS")
print("============================================")

print("\nMachine\tN\tR\tQ")

for machine in machines:

    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]:.2f}\t"
        f"{Q[machine]:.2f}"
    )


print("\nTotal Reward after 10 plays =", total_reward)

# Final best machine
best_machine = max(Q, key=Q.get)

print("Highest Q-value Machine =", best_machine)
print("Highest Q-value =", Q[best_machine])
```

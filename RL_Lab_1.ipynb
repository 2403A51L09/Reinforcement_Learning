Q1

```python
n = int(input("Enter number of machines: "))

machines = {}
q_values = {}

for i in range(n):
    name = input(f"\nEnter name of machine {i+1}: ")
    rewards = list(map(float, input("Enter rewards separated by spaces: ").split()))

    machines[name] = rewards

    q = sum(rewards) / len(rewards)
    q_values[name] = q

print("\n--- Estimated Q Values ---")

for name in machines:
    print(f"Machine {name}")
    print(f"Rewards: {machines[name]}")
    print(f"Q({name}) = {q_values[name]:.2f}")

best_machine = max(q_values, key=q_values.get)

print("\nGreedy Selected Machine:", best_machine)
```

Q2

```python
n = int(input("Enter number of machines: "))

machines = {}
Q = {}
N = {}


for i in range(n):
    name = input(f"\nEnter name of machine {i+1}: ")
    rewards = list(map(float, input("Enter rewards separated by spaces: ").split()))

    machines[name] = rewards

    Q[name] = 0
    N[name] = 0

print("\n--- Incremental Updates ---")

for name in machines:
    print(f"\nMachine {name}")

    for R in machines[name]:

        N[name] += 1

        Q[name] = Q[name] + (R - Q[name]) / N[name]

        print(f"Reward = {R}, N = {N[name]}, Q = {Q[name]:.2f}")

print("\n--- Final Q Values ---")

for name in machines:
    print(f"Q({name}) = {Q[name]:.2f}")

best_machine = max(Q, key=Q.get)

print("\nGreedy Selected Machine:", best_machine)
```
